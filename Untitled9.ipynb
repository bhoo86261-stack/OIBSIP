```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from google.colab import files
from IPython.display import clear_output # Added for clear_output function
```

```python
uploaded = files.upload()
```

```python
file_name = next(iter(uploaded))
df = pd.read_csv(file_name)
print(list(uploaded.keys()))
print(df.shape)
df.head()
```

```python

df['StateCode'] = df['customer_state'].astype('category').cat.codes

X = df[['customer_zip_code_prefix','StateCode']]
scaler = StandardScaler()
X = scaler.fit_transform(X)


```

```python
#applying K means cluster
model = KMeans(n_clusters=4, random_state=42, n_init=10)
df['Cluster'] = model.fit_predict(X)

df.head()
plt.figure(figsize=(7,5))
plt.scatter(df['customer_zip_code_prefix'], df['StateCode'], c=df['Cluster'])
plt.xlabel('ZIP Code')
plt.ylabel('State')
plt.title('Customer Segmentation')
plt.show()
print(df['Cluster'].value_counts().sort_index())
print('Total customers:', len(df))
print('Total clusters:', df['Cluster'].nunique())
```

```python
df['Cluster']
#visualize clusters
plt.figure(figsize=(7,5))
sns.scatterplot(data=df, x='customer_zip_code_prefix',
                y='StateCode', hue='Cluster')
plt.title('Customer Clusters')
plt.show()
#profile each cluster
profile = df.groupby('Cluster')[['customer_zip_code_prefix','StateCode']].mean().round(2)
print(profile)
```

```python
#number of customers per cluster
counts = df['Cluster'].value_counts().sort_index()

plt.figure(figsize=(7,4))
sns.barplot(x=counts.index, y=counts.values)
plt.xlabel('Cluster')
plt.ylabel('Number of Customers')
plt.title('Customers per Cluster')
plt.show()
```

```python
#marketing insights
for c in counts.index:
    print(f'Cluster {c}: {counts[c]} customers')
    print('Action: Use targeted regional promotions for this customer group.\n')
```
